```python
def is_even(num):
    """
    This function returns whether the given number is even or odd.

    Input: Any valid integer
    Output: 'even' if the number is even, 'odd' if the number is odd,
            'Invalid input' if input is not an integer.
    """
    if type(num) == int:
        if num % 2 == 0:
            return 'even'
        else:
            return 'odd'
    else:
        return 'Dear integer likh'
```

```python
#Acess docs
print(is_even.__doc__)
```

```python
#function_name(input)
for i in range(1,11):
    x=is_even(i)
    print(x)
```

```python
#args and kwargs
def multiply(*args):
    product=1

    for i in args:
        product=product*i

    print(args)
    return product    
```

```python
#kwargs
def display(**kwargs):
    for(key,value) in kwargs.items():
        print(key,'->',value)
```

```python
# Syntax
lambda arguments: expression

# Example: A lambda function that adds two numbers
add = lambda x, y: x + y
print(add(5, 3))  # Output: 8

# Example: Using lambda with map()
numbers = [1, 2, 3, 4, 5]
squared_numbers = list(map(lambda x: x**2, numbers))
print(squared_numbers)  # Output: [1, 4, 9, 16, 25]

# Example: Using lambda with filter()
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
print(even_numbers)  # Output: [2, 4]

# Example: Using lambda with sorted()
points = [(1, 2), (4, 1), (3, 3)]
sorted_points = sorted(points, key=lambda point: point[1])
print(sorted_points)  # Output: [(4, 1), (1, 2), (3, 3)]
```
