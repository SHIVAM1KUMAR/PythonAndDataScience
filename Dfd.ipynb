```python
arr =[3,5,7,9,11]

max=0
def is_prime(num):
    if num<=1:
        return 0
    for i in range(2,int(num/2)+1):
        if num%i==0:
           return 0
    return 1
for i in range(len(arr)):
    if is_prime(arr[i]):
        if arr[i]>max:
            max=arr[i]
print("Array",arr)  
print("max pn",max)  
```

```python
#power function
# Syntax: pow(base, exponent, mod=None)
result = pow(2, 3)  # 2^3
print("Result using pow():", result)

# With modulus
mod_result = pow(2, 3, 5)  # (2^3) % 5
print("Result with modulus:", mod_result)
```

```python
#count a word repeated in string
# Input string
string = "aaabbaacgdjk"

# Dictionary to store character counts
char_count = {}

# Loop through each character in the string
for char in string:
    if char in char_count:
        char_count[char] += 1
    else:
        char_count[char] = 1

# Print characters that are repeated
print("Repeated characters:")
for char, count in char_count.items():
    if count > 1:
        print(f"'{char}' is repeated {count} times.")
```
