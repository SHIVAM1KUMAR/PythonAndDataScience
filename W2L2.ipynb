```python
#OPERATIONS
t1=(1,2,3,4,5)
t2=(6,7,8,9,10)
print(t1+t2)
print(t1*3)
1 in t1#memebership
for i in t1:
    print(i)
```

```python
#FUNCTIONS
t=(10,20,30,40,50)
len(t)
min(t)
max(t)
print(sum(t))
sorted(t)
print(t.count(10))
print(t.index(50))

```
